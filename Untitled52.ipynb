```python
import pandas as pd
from nrclex import NRCLex

# Load your Excel dataset
df = pd.read_excel(
    "/content/cleaned_depression_dataset (4).xlsx"
)

# Define NRC sentiment and emotion function
def get_nrc_sentiment(text):
    emotion = NRCLex(str(text))
    return emotion.affect_frequencies

# Apply NRC sentiment analysis
df["nrc_sentiment"] = (
    df["processed_text"]
    .fillna("")
    .astype(str)
    .apply(get_nrc_sentiment)
)

# Display results
print(df[["processed_text", "nrc_sentiment"]].head(10))
```
